```python
def f(x, y, fl):
    if x  == y:
        return 1
    if x > y:
        return 0
    if fl:
        return f(x+1, y, 1) + f(x+2, y, 0) + f(x*2, y, 1)
    else:
        return f(x+1, y, 1) + f(x*2, y, 1)
    
f(2, 22, 1)
```
