```python
def f(x, y, fl):
    if x  == y:
        return 1
    if x > y:
        return 0
    if fl:
        return f(x+1, y, 1) + f(x+2, y, 0) + f(x*2, y, 1)
    else:
        return f(x+1, y, 1) + f(x*2, y, 1)
    
f(2, 22, 1)
```

```python
def count_programs(current, target):
    # Если вышли за границы или попали в запрещенное число 25
    if current > target or current == 25:
        return 0
    # Если достигли цели
    if current == target:
        return 1
    
    # Рекурсивно складываем количество способов для каждой команды
    return (count_programs(current + 3, target) + 
            count_programs(current * 2, target) + 
            count_programs(current * 5, target))

print(count_programs(5, 115))
```
