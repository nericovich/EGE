```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True


m, n = map(int, input().split())

groups = {}

for x in range(m, n + 1):
    if is_prime(x):
        key = tuple(sorted(str(x)))
        groups.setdefault(key, []).append(x)

for group in groups.values():
    if len(group) > 1:
        print(group)
```

```python
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

def prime_divisors(n):
    s = set()
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            if is_prime(i):
                s.add(i)
            if is_prime(n // i):
                s.add(n // i)
    return s

count = 0
n = 7500001
while count < 5:
    divs = prime_divisors(n)
    if divs:
        M = max(divs) + min(divs)
        # Исправлено n на M во втором условии
        if M % 100 == 32 and M % len(divs) == 0:
            print(n, M) # Добавил вывод M на всякий случай
            count += 1
    n += 1
```
