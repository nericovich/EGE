```python
with open('17_1.txt') as f:
    num = [int(x) for x in f.readlines()]
mini = 100001
for i in num:
    if (i % 25 == 0) and i < mini:
        mini = i
sums = []
cnt = 0
for i in range(len(num) -1):
    if num[i] % 2 ==0 or num[i+1] % 2 ==0:
        if (num[i] + num[i+1]) % mini==0:
            cnt +=1
            sums.append(num[i] + num[i+1])

max(sums)
```

```python
with open('17_3.txt') as f:
    num = [int(x) for x in f.readlines()]
maxi = -10000000
for i in num:
    if (abs(i) % 2026 == 0) and i > maxi: 
        maxi = i

sums = []
cnt = 0

for i in range(len(num) - 1):
    if num[i] * num[i+1] <0:
         if (num[i] + num[i+1]) <= maxi:
            cnt +=1
            sums.append(num[i] + num[i+1])

max(sums)
```

```python
with open('17_4.txt') as f:
    num = [int(x) for x in f.readlines()]

a = []
d = sorted(num, reverse=True)[0] * sorted(num, reverse=True)[1]
for i in range(len(num) - 2):
    if num[i] <0 or num[i+1] <0 or num[i+2] <0:
        if (num[i] * num[i+1] * num[i+2]) > 0:
            if (num[i] * num[i+1] * num[i+2]) <= d:  
                a.append(num[i] + num[i+1] + num[i+2])

len(a), max(a)
```
