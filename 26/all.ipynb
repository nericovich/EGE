```python
data = open('2626--16.txt')


n, m, k = map(int, data.pop(0).split())
occupied_places = [[0, m + 1] for i in range(k)]
for row in data:
    row_m, row_k = map(int, row.split())
    occupied_places[row_k - 1].append(row_m)
    occupied_places = [sorted(places) for places in occupied_places]
distances = []
for i, values in enumerate(occupied_places):
    for place_index in range(len(values) - 1):
        current_place = values[place_index]
        next_place = values[place_index + 1]
        distance = next_place - current_place - 2
        if distance >= 0:
            distances.append([next_place - 1, distance])
            distances.sort(key=lambda x: (-x[1], x[0]))
print(distances[0])
```

```python
from collections import defaultdict

# Читаем данные из файла
with open('26_2.txt') as f:
    n = int(f.readline())
    participants = defaultdict(list)
    
    for _ in range(n):
        # Разбиваем строку: время, id участника, id задачи
        line = f.readline().split()
        if not line: break
        
        timestamp, user_id, task_id = line[0], line[1], line[2]
        participants[user_id].append(timestamp)

# 1. Сортируем времена для каждого участника
for user_id in participants:
    participants[user_id].sort()

# 2. Находим максимальное количество решенных задач
max_tasks = max(len(times) for times in participants.values())

# 3. Ищем победителя среди тех, кто решил max_tasks
winner_id = ""
min_finish_time = "9999-99-99T99:99:99" # Инициализируем бесконечно большим временем

for user_id, times in participants.items():
    if len(times) == max_tasks:
        # Время завершения последней задачи для этого участника
        finish_time = times[-1]
        
        # Если это время раньше, чем у текущего лидера
        if finish_time < min_finish_time:
            min_finish_time = finish_time
            winner_id = user_id

print(max_tasks, winner_id)
```

```python
with open('26-15.txt') as f:
    n = int(f.readline())
    data = [ x.split() for x in f.readlines() ]

for i in range(n):
    data[i][0] = int(data[i][0])
    data[i][1] = int(data[i][1])
    data[i][2] = int(data[i][2])


data.sort(key=lambda x: x[0])

q = [[], []]
cnt1 = 0
cnt = 0

for j in range(n):
    pr, vr, no = data[j]
    q[0] = [z for z in q[0] if z > pr]
    q[1] = [z for z in q[1] if z > pr]
    
    if no == 0:

        if len(q[0]) <= len(q[1]):

            if len(q[0]) < 12:
                start = pr if len(q[0]) == 0 else q[0][-1]
                q[0].append(start + vr)
                cnt1 += 1
            else:
                cnt += 1
        else:

            if len(q[1]) < 12:
                start = pr if len(q[1]) == 0 else q[1][-1]
                q[1].append(start + vr)
            else:
                cnt += 1
    else:

        if len(q[no-1]) < 12:
            start = pr if len(q[no-1]) == 0 else q[no-1][-1]
            q[no-1].append(start + vr)
            if no == 1:
                cnt1 += 1
        else:
            cnt += 1


cnt1, cnt
```

```python
f = open('26-16.txt')
n = int(f.readline())
a = [list(map(int,s.split())) for s in f]
a.sort()
ne_obslug = 0
free = [0]*10
obslug = [0]*10
for start, time, okno in a:
    if start + 40 >= free[okno]:
        obslug[okno] += 1
        free[okno] = time + max (start, free[okno])
    else:
        ne_obslug += 1
print(max(obslug), ne_obslug)
```

```python
with open('26-17.txt') as f:
    n = int(f.readline())
    data = [ x.split() for x in f.readlines() ]

for i in range(n):
    data[i][0] = int(data[i][0])
    data[i][1] = int(data[i][1])
all=[0]*86400000
for a, b in data:
    for i in range(a, b+1):
        all[i] += 1
ans=[]
i=0
while i < 86400000:
    if all[i] >= 1:
        start = i
        while i < 86400000 and all[i] >= 1:
            i += 1
        ans.append(i-1-start)
    else:
        i += 1

print(len(ans), sum(ans))
```
