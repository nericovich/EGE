```python
f = open('26-3.txt')
k = int(f.readline())
n = int(f.readline())
a = [-1] * k
cnt  =0 
num = -1
mu = [[x, y] for x, y in [map(int, line.split()) for line in f]]
mu = sorted(mu)
for i in range(n):
    x, y  = mu[i]
    for j in range(k):
        if a[j] == -1 or a[j] <= x:
            a[j] = y+1
            cnt += 1
            num = j
            break
cnt, num+1
    
```

```python
f = open('26-4.txt')
n = int(f.readline())

a = [-1] * 100
cntleg = 0
cnt = 0

mu = []
for line in f:
    x, y, c = line.split()
    x = int(x)
    y = int(y)
    mu.append((x, y, c))

mu.sort()

for x, y, c in mu:
    fl = 1

    if c == 'A':
        for j in range(100):
            if a[j] == -1 or a[j] <= x:
                a[j] = x + y
                cntleg += 1
                fl = 0
                break

    else:
        for j in range(80, 100):
            if a[j] == -1 or a[j] <= x:
                a[j] = x + y
                fl = 0
                break

    if fl:
        cnt += 1

print(cntleg, cnt)
```
