```python
def fa(xm, ym):
    xs = 0
    ys = 0
    best = float('inf')
    for q in range(len(xm)):
        s = 0
        for w in range(len(xm)):
            s+= ((xm[w]-xm[q])**2 + (ym[w]-ym[q])**2)**0.5
        if s < best:
            best = s
            xs = xm[q]
            ys = ym[q]
    return xs, ys  

with open('27_2.txt') as f:
    d = f.readlines()

x1 = []
y1 = []
x2 =[]
y2=[]
x3 =[]
y3=[]
w = []
for s in d:
    w.append(s.split())

for j in range(len(w)):
    w[j][0] = w[j][0].replace(',', '.')
    w[j][1] = w[j][1].replace(',', '.')
for i in range(1, len(w)):
    x = float(w[i][0])
    y = float(w[i][1])

    # левый кластер
    if x < 13:
        x1.append(x)
        y1.append(y)

    # средний
    elif x < 18:
        x2.append(x)
        y2.append(y)

    # правый
    elif x < 26:
        x3.append(x)
        y3.append(y)

    # всё остальное — выбросы (игнорируем)

xc1, yc1 = fa(x1, y1)
xc2, yc2 = fa(x2, y2)
xc3, yc3 = fa(x3, y3)
import math

clusters = [
    (x1, y1, xc1, yc1),
    (x2, y2, xc2, yc2),
    (x3, y3, xc3, yc3)
]

# сортируем по количеству точек
clusters.sort(key=lambda c: len(c[0]))

# минимальный кластер
xmin, ymin, xc_min, yc_min = clusters[0]

# средний кластер
xmid, ymid, xc_mid, yc_mid = clusters[1]

# --- считаем B1 ---
B1 = 0
for i in range(len(xmin)):
    dist = math.sqrt((xmin[i] - xc_min)**2 + (ymin[i] - yc_min)**2)
    if 0 < dist <= 0.4:
        B1 += 1

# --- считаем B2 ---
B2 = 0
for i in range(len(xmid)):
    dist = math.sqrt((xmid[i] - xc_mid)**2 + (ymid[i] - yc_mid)**2)
    B2 = max(B2, dist)

print(B1, B2*10000)
```
