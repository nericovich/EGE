```python
import numpy as np
from sklearn.cluster import DBSCAN

def get_cluster_center(points):
    """
    Находит центр кластера: точку, сумма расстояний от которой 
    до всех остальных точек кластера минимальна.
    """
    min_sum_dist = float('inf')
    center = None
    
    # Вычисляем матрицу расстояний между всеми точками кластера
    # Для ускорения на больших данных можно использовать матричные операции
    from scipy.spatial.distance import cdist
    dists = cdist(points, points)
    sum_dists = dists.sum(axis=1)
    
    best_idx = np.argmin(sum_dists)
    return points[best_idx]

def solve():
    # Загрузка данных (замените названия файлов на актуальные)
    try:
        data_a = np.loadtxt('27_A_2.txt')
        data_b = np.loadtxt('27_B_2.txt')
    except:
        print("Файлы не найдены. Пожалуйста, убедитесь, что 27_A.txt и 27_B.txt в папке с кодом.")
        return

    # --- Обработка файла А ---
    db_a = DBSCAN(eps=1.0, min_samples=150).fit(data_a)
    labels_a = db_a.labels_
    unique_labels_a = set(labels_a) - {-1} # Исключаем шум
    
    clusters_a = [data_a[labels_a == i] for i in unique_labels_a]
    a1 = min(len(c) for c in clusters_a)
    
    centers_a = [get_cluster_center(c) for c in clusters_a]
    target_point = np.array([8.0, 4.5])
    a2 = sum(np.linalg.norm(center - target_point) for center in centers_a)
    
    print(f"Файл A: {a1} {int(a2 * 10000)}")

    # --- Обработка файла Б ---
    db_b = DBSCAN(eps=1.0, min_samples=150).fit(data_b)
    labels_b = db_b.labels_
    unique_labels_b = set(labels_b) - {-1}
    
    clusters_b = [data_b[labels_b == i] for i in unique_labels_b]
    # Сортируем кластеры по количеству точек для нахождения минимального и среднего
    clusters_b.sort(key=len)
    
    # B1: Кластер с минимальным количеством точек
    min_cluster_b = clusters_b[0]
    center_min_b = get_cluster_center(min_cluster_b)
    dists_min = np.linalg.norm(min_cluster_b - center_min_b, axis=1)
    # Ищем точки на расстоянии <= 0.4, не включая центр (расстояние > 0)
    b1 = np.sum((dists_min <= 0.4) & (dists_min > 0))
    
    # B2: Кластер со средним количеством точек (индекс 1 из 3)
    mid_cluster_b = clusters_b[1]
    center_mid_b = get_cluster_center(mid_cluster_b)
    dists_mid = np.linalg.norm(mid_cluster_b - center_mid_b, axis=1)
    b2 = np.max(dists_mid)
    
    print(f"Файл Б: {b1} {int(b2 * 10000)}")

if __name__ == "__main__":
    solve()
```
