```python
def fa(xm, ym):
    xs = 0
    ys = 0
    best = float('inf')
    for q in range(len(xm)):
        s = 0
        for w in range(len(xm)):
            s+= (abs(xm[w]-xm[q]) + abs(ym[w]-ym[q]))
        if s < best:
            best = s
            xs = xm[q]
            ys = ym[q]
    return xs, ys  

with open('27_Б_1.txt') as f:
    d = f.readlines()

x1 = []
x2 = []
y1 = []
y2 = []
x3 = []
y3 = []

for i in range(len(d)):
    par = [float(x) for x in d[i].split()]
    if par[0] < 10 and par[1]< 15:
        x1.append(par[0])
        y1.append(par[1])
    elif 10< par[0]< 20:
        x2.append(par[0])
        y2.append(par[1])
    elif par[0] > 23 and par[1]< 20:
        x3.append(par[0])
        y3.append(par[1]) 

xc1, yc1 = fa(x1, y1)
xc2, yc2 = fa(x2, y2)
xc3, yc3 = fa(x3, y3)
len(x1), len(x2), len(x3)
(abs(xc1-xc2)+abs(yc1-yc2))*10000

def nearest_other(xc, yc, x, y):
    best = float('inf')
    for i in range(len(x)):
        d = abs(x[i] - xc) + abs(y[i] - yc)
        if d != 0:
            best = min(best, d)
    return best

Q2 = (
    nearest_other(xc1, yc1, x1, y1) +
    nearest_other(xc2, yc2, x2, y2) +
    nearest_other(xc3, yc3, x3, y3)
)
Q2 * 10000
```
