```python
with open('24_31.txt') as f:
    s = f.readline()
lens = []
cnte=0
cnt=0
for i in range(len(s)):
    if s[i] == 'E':
        cnte=1
        cnt=1
        j=i+1
        fl=1
        while cnte<240:
            if j < len(s)-1 and s[j]=="E":
                cnte+=1
            cnt+=1
            j+=1
            if j>len(s)-1:
                fl=0
                break
        if fl:
            lens.append(cnt)
print(sorted(lens))
```

```python
with open('24_32.txt') as f:
    s = f.readline()
lens = []
i=0
cnt = ''
while i < len(s):
    if s[i] in '123456789ABCD':
        cnt += s[i]
        for j in range(i+1, len(s)):
            if s[j] in  '0123456789ABCD':
                cnt+= s[j]
            else:
                if len(cnt)>1:
                    for k in range(len(cnt)-1, 0, -1):
                        if s[k] in '02468AC':
                            lens.append(len(cnt[0:k+1]))
                
                cnt=''
                i=j+1

    i+=1

sorted(lens, reverse=True)
```
